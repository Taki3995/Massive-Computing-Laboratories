## Functions needed for Benchmark notebook

```python
def work(task):
    """
    Some amount of work that will take time
    
    Parameters
    ----------
    task : tuple
        Contains number, loop, and number processors
    """
    number, loop = task
    b = 2. * number - 1.
    for i in range(loop):
        a, b = b * i, number * i + b
    return a, b
```
